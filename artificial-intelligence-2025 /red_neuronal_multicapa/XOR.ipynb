```python
import numpy as np

# Datos XOR
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([[0],
              [1],
              [1],
              [0]])

# valores de la tabla de verdad para XOR
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

modelo = Sequential([
    Dense(4, activation='relu', input_dim=2),
    Dense(1, activation='sigmoid')
])
#se manda llamar la clase(plantilla) "Sequential" desde el módulo "models" de ubicado en el submódulo "keras" de la librería tensorflow.
#Este módulo nos ayuda a setear el módelo de red neuronal que usaremos, en este caso seteamos 4 neuronas y estamos diciendole que hay dos entradas para cada muestra.
#para la clase "Desnse" es la que enlaza las neuronas en su totalidad una con otra y lo seteamos para que nos devuelva una sola salida de dato binario.
```

```python
from tensorflow.keras.optimizers import Adam

# Tasa de aprendizaje deseada
learning_rate = 0.01
adam_optimizer = Adam(learning_rate=learning_rate)

#la clase "Adam" es un algoritmo que ajusta los pesos de forma automatica durante el aprendizaje y minimizar la pérdida
```

```python
modelo.compile(
    optimizer=adam_optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)

#aquí se realiza la configurazión del modelo que se utilizará
```

```python
history = modelo.fit(
    X, y,
    epochs=50,
    verbose=0,
)

#se definen la epocas que queremos que el modelo realice para que aprenda
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['loss'])
plt.title('Evolución de la pérdida (Loss)')
plt.xlabel('Época')
plt.ylabel('Binary Cross-Entropy')
plt.show()

#usamos la librería de "matplotlib" para raficar y le damos las características de la gráfica.
```

```python
loss, accuracy = modelo.evaluate(X, y)
print(f'Loss: {loss}, Accuracy: {accuracy}')

#medimos el desempeño que ha tenido el modelo y le pedímos que lo muestre
```

```python
pred = (modelo.predict(X) > 0.5).astype(int)
for i, (inp, p) in enumerate(zip(X, pred)):
    print(f"{inp} -> {p[0]}")

#vertimos las salidas y le pedimos que imprima el resultado final del modelo.
```
